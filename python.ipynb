Operators Python Problems


```python
# Program to calculate nth term of an arithmetic progression
a = 5000  # First term
d = 2000  # Common difference
n = 61    # Term number

# Calculate the nth term
nth_term = a + (n - 1) * d

# Output the result
print("The 61st term of the arithmetic progression is:", nth_term)
```

```python
# Program to calculate nth term of a geometric progression
# Input the first term (a)
a = 5000  	# First term
r = 2  		# Common ratio
n = 11    	# Term number

# Calculate the nth term using the formula
nth_term = a * (r ** (n - 1))

# Print the result
print("The 11th term of the geometric progression is: ", nth_term)
```

```python
# Program to calculate the sum of the first n natural numbers
# Input the value of n
n = 10
# Calculate the sum using the formula
sum_n = n * (n + 1) // 2
# Print the result
print("The sum of the first", n, "natural numbers is:", sum_n)
```

```python
# Program to find the last digit of a non-negative number

x = 234

# Find the last digit
last_digit = x % 10

# Print the result
print("The last digit is:", last_digit)
```

```python
d = 0
n = 9

# Calculate the day before `n` days
day_before_n_days = (d - n) % 7

# Print the result
print("The day", n, "days before day", d, "is:", day_before_n_days)
```

```python
#Using the dict for the calculator problem

class Solution:
  def calculate(self, a : int, b : int,  operator : int):
    operation = {
        1 : a+b,
        2: a-b,
        3: a*b,
        4: int(a/b)
    }
    if operator in operation:
      return operation[operator]
    else:
      return "invalid"
a = int(input())
b = int(input())
operator = int(input())
res = Solution()
print(res.calculate(a,b,operator))
```

```python
# square star
n = int(input("enter a num: "))
for i in range(n):
  for j in range(n):
    print("*",end=" ")
  print("")
```

```python
# half triangle
n = int(input("enter a num: "))
for i in range(n):
  for j in range (i+1):
    print("*",end=" ")
  print("")
```

```python
# inverted half triangle
n = int(input("enter a num: "))
for i in range(n):
  for j in range (n-i):
    print("*",end=" ")
  print("")
```

```python
# triangle
n = int(input("enter a num: "))
for i in range(n):
  for j in range(n-i-1):
    print(' ',end ='')
  for k in range(2*i + 1):
    print("*" , end='')
  print()
```

```python
# inverted triangle
n = int(input("enter a num: "))
for i in range(n,0,-1):
  for j in range(n-i):
    print(' ', end = '')
  for k in range(2*i - 1):
    print('*', end = '')
  print()
```

```python
# factorial
result = 1
n = 4
for i in range (2,n+1):
  result *= i
print(result)

#using built in function

import math
n = 4
res = math.factorial(n)
print(res)
```

```python
# gcd
a = int(input("enter a num: "))
b = int(input("enter a num: "))
small = min(a,b)
gcd = 1
for i in range (1,small + 1):
  if a % i == 0 and b % i == 0:
    gcd = i
print(gcd)

# built in function

import math
a = 4
b = 6
res = math.gcd(a,b)
print(res)
```
